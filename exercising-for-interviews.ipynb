**Question:**

*write a function that returns the first character of a string which is not repeated*

e.g. 

first_unique_char("swiss")       # "w"

first_unique_char("aabbcc")      # None

first_unique_char("python")      # "p"

first_unique_char("aabbcddee")   # "c"


```python
# I intuitevily solved the question thinking to an iteration check for each character. If not repeated, the character is printed.

def first_unique_char(word):
    for i in word:
        if word.count(i)==1:
            return i
    return None
```

*However as I learnt this approach could elicit iteration through all characters: this, in long words, could mean quadratic time complexity... 
Under suggestion the approach safer for time consuption requires to create a "box" (a dictionary in python) to count the number of characters in a word* 

```python
def first_unique_char(word):
    #the dictionary is called counts
    counts = {}
    
    for char in word:
        counts[char] = counts.get(char, 0) + 1
    #I add the print as a check of what this first part does
        print (counts)
```

*Nice. In a second part, programm will check for the first character not ripeated (it should be counted only once in a given word)*

```python
def first_unique_char(word):
    #the dictionary is called counts
    counts = {}
    
    for char in word:
        counts[char] = counts.get(char, 0) + 1
    #removing the check print and adding the "search for uniqueness" (or count of character equals one)       
        if counts[char] == 1:
            return char
    return None
```

```python
first_unique_char("canada")
```

The dictionary allows separation of operations:
one is used to count the characters in the word; the other checks if those counts equal the target (=1).
This reduced the operation time to a linear form, thus avoiding the potential quadratic unwanted timing.
